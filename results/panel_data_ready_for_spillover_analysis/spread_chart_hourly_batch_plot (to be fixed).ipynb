```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import warnings
```

```python
def plot_event_chart(df, n_hours, output_dir):
    
    # iterate over each event column
    for col in df.columns[1:]:  # skip the first column "Time"
        time = df["Time"]
        abnormal_spread = df[col]

        mask_before_0 = time < 0 and time >= -24  # 24 hours before the event
        mask_after_0 = time >= 0                  # 0 hour and 10 hours afterward
        
        time_before_0 = time[mask_before_0]
        spread_before_0 = abnormal_spread[mask_before_0]
        
        time_after_0 = time[mask_after_0]
        spread_after_0 = abnormal_spread[mask_after_0]

        # 计算累积异常扩散值（从 0 时点开始）
        accumulative_spread = spread_after_0.cumsum()

        # plot the main y-axis (spread before the 0 hour and abnormal spread after the 0 hour)
        plt.figure(figsize=(16, 9))
        fig, ax1 = plt.subplots()
        ax1.plot(time_before_0, spread_before_0, color="green", label="Spread", alpha=0.7)
        ax1.plot(time_after_0, spread_after_0, color="blue", label="Abnormal Spread", alpha=0.7)

        # plot the secondary y-axis (accumulative abnormal spread after the 0 hour)
        ax2 = ax1.twinx()  # create a second y-axis sharing the same x-axis
        ax2.plot(time_after_0, accumulative_spread, color="red", label="Accumulative AS", alpha=0.7)
        ax2.set_ylabel("Accumulative AS", color="red")
        ax2.tick_params(axis="y", labelcolor="red")

        # add reference lines
        ax1.axhline(0, color="gray", linestyle="--", linewidth=0.8)  # horizontal line
        ax1.axvline(0, color="black", linestyle="--", linewidth=1, label="Event Occurrence")  # vertical line

        # set widets
        ax1.set_xlabel("Time (hours)")
        ax1.set_ylabel("Spread")
        fig.suptitle(f"{n_hours} hours {col} Spread Chart")
        ax1.legend(loc="upper left")
        ax2.legend(loc="upper right")

        # save the chart
        output_path = os.path.join(output_dir, f"{col}_{}.png")
        plt.savefig(output_path, dpi=72)
        plt.close()

    print(f"Saved all charts to {output_dir}")
```

```python
for  in :
    output_dir = f"event_charts_{i}_hours"
    os.makedirs(output_dir, exist_ok=True)
    file_path = f"panel_df_{i}_hours.csv"
    df = pd.read_csv(file_path)
    plot_event_chart(df, i, output_dir)
```
